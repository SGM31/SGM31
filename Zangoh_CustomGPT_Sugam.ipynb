```python
#First we will scrap the article and prepare a text file.
#Article Link - https://indianexpress.com/article/cities/delhi/bangalore-bound-flight-forced-to-return-to-delhi-after-air-conditioning-unit-catches-fire-mid-air-9335736/
```

```python
# Installing libraries
!pip install requests beautifulsoup4

# Import required libraries
import requests
from bs4 import BeautifulSoup

# Function to fetch and clean the article content
def fetch_article_content(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    # Extracting the article's text
    article_text = ' '.join([p.text for p in soup.find_all('p')])
    return article_text

# URL of the article
article_url = 'https://indianexpress.com/article/cities/delhi/bangalore-bound-flight-forced-to-return-to-delhi-after-air-conditioning-unit-catches-fire-mid-air-9335736/'

# Fetch and clean the article content
article_content = fetch_article_content(article_url)

# Save article content to a file
with open('article_content.txt', 'w') as f:
    f.write(article_content)

# Output the first few lines of the file to verify
with open('article_content.txt', 'r') as f:
    for _ in range(1):
        print(f.readline().strip())
```

```python
#All Lines print of article
with open('article_content.txt', 'r', encoding='utf-8') as f:
    all_lines = f.readlines()

for line in all_lines:
    print(line.strip())
```

```python
## Last line is not from the article but from a side panel of news opened with the article link. Truncating them we get original file
with open('article_content.txt', 'r', encoding='utf-8') as f:
    content = f.read()
words = content.split()
# Truncate the last 20 words(counted the words maunally due to short article length)
truncated_content = ' '.join(words[:-20])
print(truncated_content)
#final text data from article
```
